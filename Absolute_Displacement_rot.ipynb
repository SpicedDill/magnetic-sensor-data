```python
# Initial magnet position (mm)
p0 = np.array([30, 0, -0.5])

# Rotation angles (degrees)
yaw = 3.0
pitch = 0.0
roll = 0.0

# Convert degrees to radians
yaw = np.deg2rad(yaw)
pitch = np.deg2rad(pitch)
roll = np.deg2rad(roll)

# Rotation matrices
Rx = np.array([
    [1, 0, 0],
    [0, np.cos(roll), -np.sin(roll)],
    [0, np.sin(roll),  np.cos(roll)]
])

Ry = np.array([
    [ np.cos(pitch), 0, np.sin(pitch)],
    [0,              1, 0],
    [-np.sin(pitch), 0, np.cos(pitch)]
])

Rz = np.array([
    [np.cos(yaw), -np.sin(yaw), 0],
    [np.sin(yaw),  np.cos(yaw), 0],
    [0,            0,           1]
])

# Apply roll -> pitch -> yaw
p_final = Rz @ Ry @ Rx @ p0

# Displacement vector
displacement = p_final - p0

# Total displacement magnitude
total_displacement = np.linalg.norm(displacement)

print("Initial position (mm):", p0)
print("Final position (mm):  ", p_final)
print("Displacement (mm):    ", displacement)
print("Total displacement:    ", total_displacement, "mm")
```
